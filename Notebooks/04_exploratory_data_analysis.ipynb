# Exploratory Data Analysis

EDA is conducted to understand customer behavior and identify factors associated with customer churn.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../Data/processed/telecom_churn_cleaned.csv")
```

```python
sns.countplot(x='churn', data=df)

plt.title("Customer Churn Distribution")
plt.show()
```

### Churn Distribution

The majority of customers did not churn, indicating a class imbalance in the dataset.

This imbalance should be considered during model training and evaluation.

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

### Correlation Analysis

Strong correlations exist between call minutes and corresponding charges because charges are calculated directly from usage.

Understanding these relationships helps identify redundant features and important predictors.

```python
sns.boxplot(
    x='churn',
    y='customer service calls',
    data=df
)

plt.show()
```

### Customer Service Calls vs Churn

Customers who churn tend to make more customer service calls.

This suggests that customer dissatisfaction may be a strong contributor to churn behavior.

```python
sns.countplot(
    x='international plan',
    hue='churn',
    data=df
)

plt.show()
```

### International Plan Analysis

Customers subscribed to international plans appear to exhibit different churn patterns compared to those without international plans.
